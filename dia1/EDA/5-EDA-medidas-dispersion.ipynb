# MEDIDAS DE DISPERSIÓN

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
sns.set_style(style='whitegrid')
sns.set_context(context='notebook')
plt.rcParams['figure.figsize'] = (11, 9.4)

penguin_color = {
    'Adelie': '#ff6602ff',
    'Gentoo': '#0f7175ff',
    'Chinstrap': '#c65dc9ff'
}

penguins_df = sns.load_dataset('penguins')
penguins_df.dropna(inplace=True)
penguins_df
```

# RANGO

## MAXIMO

```python
penguins_df.max(numeric_only=True)
```

# MINIMO

```python
penguins_df.min(numeric_only=True)
```

# RANGO = MAX - MIN

```python
rango = penguins_df.max(numeric_only=True) - penguins_df.min(numeric_only=True)
rango
```

# DESVIACIÓN STANDARD

```python
penguins_df.std(numeric_only=True)
```

# RANGO INTECUARTILICO

# Q1 = 25%

```python
penguins_df.quantile(0.25,numeric_only=True)
```

# Q3 = 75%

```python
penguins_df.quantile(0.75,numeric_only=True)
```

# RANGO DE Q1 - Q3

```python
rango_intercuartilico = penguins_df.quantile(0.75,numeric_only=True) - penguins_df.quantile(0.25,numeric_only=True)
rango_intercuartilico
```

# MOSTRAMOS TODOS LOS DATOS

```python
(
    penguins_df
    .quantile(q=[0.75,0.50,0.25],numeric_only=True)
    .transpose()
    .rename_axis('metric')
    .reset_index()
    .assign(
        iqr = lambda df: df[0.75] - df[0.25]
    )
)
```

```python
variable = 'body_mass_g'

sns.histplot(
    data=penguins_df,
    x=variable
)

# MEDIA
plt.axvline(
    x=penguins_df[variable].mean(),
    label='Media',
    color='red',
    linestyle='dashed',
    linewidth=2
)

# MEDIANA
plt.axvline(
    x=penguins_df[variable].median(),
    label='Mediana',
    color='green',
    linestyle='dashed',
    linewidth=2
)

#MODA
plt.axvline(
    x=penguins_df[variable].mode().values[0],
    label='Moda',
    color='yellow',
    linestyle='dashed',
    linewidth=2
)


#Q 25
plt.axvline(
    x=penguins_df[variable].quantile(0.25),
    label='Q1',
    color='black',
    linestyle='dashed',
    linewidth=2
)

#Q 75
plt.axvline(
    x=penguins_df[variable].quantile(0.75),
    label='Q3',
    color='black',
    linestyle='dashed',
    linewidth=2
)
plt.legend()

# std
plt.axvline(
    x=penguins_df[variable].mean() + penguins_df[variable].std(),
    label='std',
    color='blue',
    linestyle='dashed',
    linewidth=2
)

plt.show()
```

# MOSTRANDO RANGO Y VALORES ATIPICOS

```python
sns.boxplot(x=penguins_df['body_mass_g'], color='#0f7175ff')
plt.title('Distribución general del peso corporal de los pingüinos')
plt.xlabel('Body Mass (g)')
plt.show()
```

# DETECTANDO VALORES ATIPICOS

```python
# Calcular límites del IQR
q1 = penguins_df['body_mass_g'].quantile(0.25)
q3 = penguins_df['body_mass_g'].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

# Visualizar boxplot
sns.boxplot(x=penguins_df['body_mass_g'], color='#c65dc9ff')
plt.axvline(lower_bound, color='red', linestyle='--', label='Límite inferior')
plt.axvline(upper_bound, color='red', linestyle='--', label='Límite superior')
plt.title('Detección de valores atípicos en el peso corporal')
plt.xlabel('Body Mass (g)')
plt.legend()
plt.show()

# Mostrar valores atípicos
outliers = penguins_df[
    (penguins_df['body_mass_g'] < lower_bound) |
    (penguins_df['body_mass_g'] > upper_bound)
]
outliers
```
